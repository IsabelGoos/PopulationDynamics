## Building the model

The exponential growth model can be derived by considering a population of unicellular microorganisms that reproduce through cellular division. To simplify the model, we assume the following:

- **Population size:** The population is sufficiently large to be represented as a continuous variable.
- **Growth factor:** Population growth depends only on time.
- **Environmental conditions:** The population has access to sufficient nutrients, and environmental conditions, such as temperature, remain constant over time.
- **Reproduction:** Each microorganism reproduces through cellular division, producing two new microorganisms.
- **Mortality:** Death due to external factors, such as nutrient depletion, is not considered. A microorganism ceases to exist only when it divides into two new microorganisms.


Let $$P(t)$$ denote the **population size at time t**.

Let's assume that all organisms reproduce at the same constant rate, denoted by $$K > 0,$$ which represents the **reproduction rate (per unit of time) per organism**. Thus, $K$ describes the rate at which the population grows relative to its current size.


By measuring the population at the two points in time $t$ and $t+\Delta t$, with $\Delta t$ small, we obtain the following approximation:

$$P(t+\Delta t)-P(t) \approx KP(t)\Delta t.$$

Dividing both sides by $\Delta t$ and taking the limit as $\Delta t \to 0$, we obtain the following autonomous ordinary differential equation, known as **Malthus' Law of Population Growth**:

$$\frac{\text{d} P}{\text{d} t} =  KP.$$


## Solving the model

This differential equation is straightforward to solve. We are looking for a function whose derivative is proportional to itself. The exponential function satisfies precisely this property, giving the **general solution**:

$$P(t) = C e^{Kt},$$

where $C$ is an arbitrary constant determined by the initial condition.


Furthermore, we can show that every solution must be of this form by showing that the derivative of $P(t)e^{-Kt}$ is zero:

\begin{aligned}
\frac{\text{d}}{\text{d}t}\left(P(t)e^{-Kt}\right)
&= P'(t)e^{-Kt} - KP(t)e^{-Kt}\\
&= KP(t)e^{-Kt} - KP(t)e^{-Kt}\\
&= 0.
\end{aligned}


Since its derivative is zero, $P(t)e^{-Kt}$ must be constant. 


Finally, if $$P(0) = P_0$$ is the **initial population size**, the associated solution is $$P(t) = P_0 e^{Kt}.$$


## Stability Analysis

The stability analysis depends on both the model parameter $K$ and the initial condition $P_0$, although they play different roles.


**1. The parameter $K$: determines the long-term dynamics**

The sign of $K$ determines whether the population grows, remains constant, or decays.

| Parameter | Long-term behavior |
|---|---|
| $K > 0$ | Exponential growth |
| $K = 0$ | Constant population |
| $K < 0$ | Exponential decay towards zero |


**2. The initial condition $P_0$: determines the starting point**

The initial value determines where the solution starts.

- $P_0 > 0$: Positive initial population.
- $P_0 = 0$: Population remains zero for all time.
- $P_0 < 0$: Mathematically valid solution, but generally not biologically meaningful.


**3. Equilibrium Point**

An equilibrium point is a population value $P^*$ for which the constant function $P(t) \equiv P^*$ is a solution of the differential equation. In other words, the population remains unchanged over time, meaning that:

$$
P'(t)=0
$$

For the Malthusian model, this gives:

We can examine what happens when the population is slightly perturbed away from this equilibrium:

- **If $K < 0$:** Solutions approach zero regardless of their initial value. The equilibrium is asymptotically stable.
- **If $K > 0$:** Any nonzero initial perturbation moves away from zero. The equilibrium is unstable.
- **If $K = 0$:** Solutions remain wherever they start. The equilibrium is stable, but not asymptotically stable.





**Key distinction:**

- The parameter $K$ determines the stability properties of the equilibrium.
- The initial condition $P_0$ determines which particular solution (trajectory) we observe.

Stability is therefore a property of the system's dynamics around an equilibrium, rather than simply whether a particular solution increases or decreases.

```python

import numpy as np
import matplotlib.pyplot as plt

# Initial population
P0 = 10

# Time interval
t = np.linspace(0, 15, 500)

# Growth parameters for each scenario
scenarios = [
    {"K": 0.3, "title": "Exponential Growth (K > 0)"},
    {"K": 0.0, "title": "Constant Population (K = 0)"},
    {"K": -0.3, "title": "Exponential Decay (K < 0)"}
]

# Create one subplot for each scenario
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, scenario in zip(axes, scenarios):
    K = scenario["K"]
    P = P0 * np.exp(K * t)

    ax.plot(t, P, linewidth=2)
    ax.set_title(scenario["title"])
    ax.set_xlabel("Time (t)")
    ax.set_ylabel("Population P(t)")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
